```python
#!/usr/bin/env python
# coding: utf-8
from __future__ import annotations

import argparse
import ast
import csv
import re
from pathlib import Path
from typing import Any, Dict, List


STEP_RE = re.compile(r"^Step\s+(?P<step>\d+):\s*(?P<lhs>.*?)\s*=\s*(?P<rhs>.+?)\s*$")
SEL_RE = re.compile(r"^Selection:\s*(?P<metric>[A-Za-z0-9_]+)\s*\[(?P<split>[A-Za-z0-9_]+)\]\s*$")
RES_RE = re.compile(r"^Results\s*\((?P<split>[A-Za-z0-9_]+)\):\s*$")
KV_RE = re.compile(r"^(?P<key>[A-Za-z0-9_]+)\s*=\s*(?P<val>.+?)\s*$")
NUM_RE = re.compile(r"^[+\-]?(?:\d+\.?\d*|\.\d+)(?:[eE][+\-]?\d+)?$")


def _clean_box_line(line: str) -> str:
    s = line.rstrip("\n")
    if s.startswith("|") and s.endswith("|"):
        s = s[1:-1]
    return s.strip()


def _parse_value(text: str) -> Any:
    t = text.strip()
    if t.endswith("*"):
        t = t[:-1].strip()
    try:
        return ast.literal_eval(t)
    except Exception:
        pass
    if NUM_RE.match(t):
        try:
            return float(t)
        except Exception:
            return t
    return t


def _to_number(v: Any) -> float | None:
    if isinstance(v, (int, float)):
        return float(v)
    if isinstance(v, str) and NUM_RE.match(v.strip()):
        try:
            return float(v)
        except Exception:
            return None
    return None


def _parse_step_params(lhs: str, rhs: str) -> Dict[str, Any]:
    params: Dict[str, Any] = {}
    names = [x.strip() for x in lhs.split("+")] if "+" in lhs else [lhs.strip()]
    rhs_val = _parse_value(rhs)

    if len(names) == 1:
        params[names[0]] = rhs_val
        return params

    if isinstance(rhs_val, tuple) and len(rhs_val) == len(names):
        for k, v in zip(names, rhs_val):
            params[k] = v
        return params

    params[lhs.strip()] = rhs_val
    return params


def parse_overrides(path: Path) -> List[Dict[str, Any]]:
    lines = path.read_text(encoding="utf-8", errors="replace").splitlines()
    records: List[Dict[str, Any]] = []

    global_metric = "MF1"
    global_split = "avg"
    for line in lines:
        m = SEL_RE.match(line.strip())
        if m:
            global_metric = m.group("metric").upper()
            global_split = m.group("split").lower()
            break

    cur: Dict[str, Any] | None = None
    mode: str | None = None
    cur_split: str | None = None

    def flush() -> None:
        nonlocal cur
        if cur is not None:
            records.append(cur)
            cur = None

    for raw in lines:
        line = _clean_box_line(raw)
        if not line:
            continue

        m_step = STEP_RE.match(line)
        if m_step:
            flush()
            step = int(m_step.group("step"))
            lhs = m_step.group("lhs")
            rhs = m_step.group("rhs")
            cur = {
                "step": step,
                "label": lhs,
                "raw_value": rhs,
                "params": _parse_step_params(lhs, rhs),
                "fixed": {},
                "metrics": {"dev": {}, "test": {}, "avg": {}},
                "selection_metric": global_metric,
                "selection_split": global_split,
            }
            mode = None
            cur_split = None
            continue

        if cur is None:
            continue

        m_sel = SEL_RE.match(line)
        if m_sel:
            cur["selection_metric"] = m_sel.group("metric").upper()
            cur["selection_split"] = m_sel.group("split").lower()
            continue

        if line == "Fixed:":
            mode = "fixed"
            cur_split = None
            continue

        m_res = RES_RE.match(line)
        if m_res:
            mode = "results"
            cur_split = m_res.group("split").lower()
            cur["metrics"].setdefault(cur_split, {})
            continue

        m_kv = KV_RE.match(line)
        if not m_kv:
            continue
        k = m_kv.group("key")
        v = _parse_value(m_kv.group("val"))

        if mode == "fixed":
            cur["fixed"][k] = v
        elif mode == "results" and cur_split is not None:
            cur["metrics"][cur_split][k.upper()] = v

    flush()

    for r in records:
        merged = dict(r.get("fixed", {}))
        merged.update(r.get("params", {}))
        r["params_merged"] = merged

        sel_split = str(r.get("selection_split", "avg")).lower()
        sel_metric = str(r.get("selection_metric", "MF1")).upper()
        selected = r.get("metrics", {}).get(sel_split, {}).get(sel_metric)
        r["selected_score"] = _to_number(selected)

    return records


def _col_value(rec: Dict[str, Any], col: str) -> Any:
    if col in {"step", "label", "selection_split", "selection_metric", "selected_score"}:
        return rec.get(col)
    if col.startswith("param:"):
        return rec.get("params_merged", {}).get(col[6:])
    if col.startswith("fixed:"):
        return rec.get("fixed", {}).get(col[6:])
    if "." in col:
        split, metric = col.split(".", 1)
        return rec.get("metrics", {}).get(split.lower(), {}).get(metric.upper())
    return rec.get("params_merged", {}).get(col)


def _match_filters(rec: Dict[str, Any], where: List[str]) -> bool:
    for expr in where:
        if "=" not in expr:
            return False
        key, raw_val = expr.split("=", 1)
        key = key.strip()
        want = _parse_value(raw_val.strip())
        got = _col_value(rec, key)
        if str(got) != str(want):
            return False
    return True


def _as_text(v: Any, max_width: int) -> str:
    s = "" if v is None else str(v)
    if max_width > 0 and len(s) > max_width:
        return s[: max_width - 1] + "…"
    return s


def _print_table(rows: List[Dict[str, Any]], columns: List[str], max_width: int) -> None:
    headers = columns
    matrix: List[List[str]] = [[_as_text(_col_value(r, c), max_width=max_width) for c in columns] for r in rows]
    widths = [len(h) for h in headers]
    for row in matrix:
        for i, cell in enumerate(row):
            widths[i] = max(widths[i], len(cell))

    def sep() -> str:
        return "+-" + "-+-".join("-" * w for w in widths) + "-+"

    print(sep())
    print("| " + " | ".join(h.ljust(widths[i]) for i, h in enumerate(headers)) + " |")
    print(sep())
    for row in matrix:
        print("| " + " | ".join(row[i].ljust(widths[i]) for i in range(len(headers))) + " |")
    print(sep())


def main() -> None:
    parser = argparse.ArgumentParser(
        description="Analyze overrides.txt from multimodal search/single run.",
        formatter_class=argparse.ArgumentDefaultsHelpFormatter,
    )
    parser.add_argument("--file", required=True, help="Path to overrides.txt")
    parser.add_argument("--metric", default=None, help="Metric to sort by (e.g. MF1, UAR, MF1_AVG)")
    parser.add_argument("--split", default=None, help="Split for metric: dev|test|avg")
    parser.add_argument("--sort-by", default=None, help="Column for sorting (e.g. avg.MF1, param:optimizer)")
    parser.add_argument("--top", type=int, default=20, help="Show top-N rows after filtering/sorting")
    parser.add_argument("--where", action="append", default=[], help="Equality filter, repeatable. Example: param:optimizer=adamw or avg.MF1=0.8125")
    parser.add_argument("--min-score", type=float, default=None, help="Min threshold for selected score")
    parser.add_argument("--max-score", type=float, default=None, help="Max threshold for selected score")
    parser.add_argument(
        "--columns",
        default="step,selected_score,selection_split,selection_metric,param:optimizer,param:lr,param:weight_decay,param:d_model,param:x_layers,param:x_heads,avg.MF1,avg.UAR,dev.MF1,test.MF1",
        help="Comma-separated output columns",
    )
    parser.add_argument("--out-csv", default=None, help="Optional output CSV path")
    parser.add_argument("--max-col-width", type=int, default=28, help="Max rendered column width in terminal")
    args = parser.parse_args()

    path = Path(args.file)
    if not path.exists():
        raise FileNotFoundError(path)

    records = parse_overrides(path)
    if not records:
        print("No step records found.")
        return

    split = (args.split or records[0].get("selection_split") or "avg").lower()
    metric = (args.metric or records[0].get("selection_metric") or "MF1").upper()
    default_sort_col = f"{split}.{metric}"
    sort_col = args.sort_by or default_sort_col

    filtered = []
    for r in records:
        if args.where and not _match_filters(r, args.where):
            continue
        score = _to_number(_col_value(r, sort_col))
        if args.min_score is not None and (score is None or score < args.min_score):
            continue
        if args.max_score is not None and (score is None or score > args.max_score):
            continue
        filtered.append(r)

    def score_key(rec: Dict[str, Any]) -> float:
        v = _to_number(_col_value(rec, sort_col))
        return float("-inf") if v is None else float(v)

    filtered.sort(key=score_key, reverse=True)
    shown = filtered[: max(0, args.top)]

    columns = [c.strip() for c in args.columns.split(",") if c.strip()]
    if not columns:
        columns = ["step", "selected_score"]

    print(f"File: {path}")
    print(f"Parsed steps: {len(records)}")
    print(f"Filtered: {len(filtered)}")
    print(f"Sort by: {sort_col}")
    if args.where:
        print(f"Filters: {args.where}")
    print("")
    _print_table(shown, columns=columns, max_width=max(8, args.max_col_width))

    if args.out_csv:
        out_path = Path(args.out_csv)
        out_path.parent.mkdir(parents=True, exist_ok=True)
        with out_path.open("w", encoding="utf-8", newline="") as f:
            writer = csv.writer(f)
            writer.writerow(columns)
            for rec in shown:
                writer.writerow([_col_value(rec, c) for c in columns])
        print(f"\nSaved CSV: {out_path}")


if __name__ == "__main__":
    main()
```
